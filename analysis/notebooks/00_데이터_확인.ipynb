## csv 불러오기 및 데이터 확인


```python
# 라이브러리 불러오기
import pandas as pd
import numpy as np
import os
# 데이터 경로: ../데이터셋 우선, 없으면 ../data
DATA_DIR = "../데이터셋" if os.path.isdir("../데이터셋") else "../data"
print(f"라이브러리 불러오기 완료")
print(f"데이터 경로: {DATA_DIR}")
```

```python
# 1. portfolio.csv 불러오기 (프로모션/오퍼 정보)
portfolio = pd.read_csv(os.path.join(DATA_DIR, "portfolio.csv"))
print(f"=== Portfolio 데이터 ===")
print(f"Shape: {portfolio.shape}")
print(portfolio.head())
print(portfolio.info())
```

```python
# 2. profile.csv 불러오기 (고객 프로필 정보)
profile = pd.read_csv(os.path.join(DATA_DIR, "profile.csv"))
print(f"=== Profile 데이터 ===")
print(f"Shape: {profile.shape}")
print(profile.head())
print(profile.info())
```

```python
# 3. transcript.csv 불러오기 (거래/이벤트 정보)
transcript = pd.read_csv(os.path.join(DATA_DIR, "transcript.csv"))
print(f"=== Transcript 데이터 ===")
print(f"Shape: {transcript.shape}")
print(transcript.head())
print(transcript.info())
```

## 결측치 확인


```python
# 1. Portfolio 데이터 결측치 확인
print("=== Portfolio 결측치 ===")
missing_portfolio = portfolio.isnull().sum()
print(missing_portfolio)
print(f"\n총 결측치 개수: {missing_portfolio.sum()}")
print(f"결측치가 있는 컬럼: {missing_portfolio[missing_portfolio > 0].index.tolist()}")
```

```python
# 2. Profile 데이터 결측치 확인
print("=== Profile 결측치 ===")
missing_profile = profile.isnull().sum()
missing_profile_pct = (profile.isnull().sum() / len(profile) * 100).round(2)
missing_df = pd.DataFrame({
    '결측치 개수': missing_profile,
    '결측치 비율(%)': missing_profile_pct
})
print(missing_df[missing_df['결측치 개수'] > 0])
print(f"\n총 결측치 개수: {missing_profile.sum()}")
print(f"결측치가 있는 컬럼: {missing_profile[missing_profile > 0].index.tolist()}")
```

```python
# 3. Transcript 데이터 결측치 확인
print("=== Transcript 결측치 ===")
missing_transcript = transcript.isnull().sum()
print(missing_transcript)
print(f"\n총 결측치 개수: {missing_transcript.sum()}")
print(f"결측치가 있는 컬럼: {missing_transcript[missing_transcript > 0].index.tolist()}")
```

## 이상치 확인


```python
# 1. Portfolio 데이터 이상치 확인
print("=== Portfolio 이상치 확인 ===")
numeric_cols = portfolio.select_dtypes(include=[np.number]).columns.tolist()
numeric_cols = [col for col in numeric_cols if col != 'Unnamed: 0']  # Unnamed: 0 제외

for col in numeric_cols:
    print(f"\n[{col}]")
    print(portfolio[col].describe())
    print(f"고유값: {sorted(portfolio[col].unique())}")
    print(f"최소값: {portfolio[col].min()}, 최대값: {portfolio[col].max()}")
```

```python
# 2. Profile 데이터 이상치 확인
print("=== Profile 이상치 확인 ===")
numeric_cols = ['age', 'income']  # became_member_on은 날짜 형태이므로 제외

for col in numeric_cols:
    if col in profile.columns:
        print(f"\n[{col}]")
        print(profile[col].describe())
        print(f"최소값: {profile[col].min()}, 최대값: {profile[col].max()}")
        print(f"고유값 개수: {profile[col].nunique()}")
        
        # age의 경우 특이값 확인 (118 같은 명백히 이상한 값)
        if col == 'age':
            print(f"\nage=118인 행 개수: {(profile[col] == 118).sum()}")
            print(f"age < 18 또는 age > 100인 행 개수: {((profile[col] < 18) | (profile[col] > 100)).sum()}")
        
        # income의 경우 특이값 확인
        if col == 'income':
            print(f"\nincome < 0인 행 개수: {(profile[col] < 0).sum()}")
            print(f"income이 매우 큰 값 (예: > 200000)인 행 개수: {(profile[col] > 200000).sum()}")
        
        # 상위/하위 값 분포 확인
        print(f"\n하위 10개 값: {sorted(profile[col].dropna().unique())[:10]}")
        print(f"상위 10개 값: {sorted(profile[col].dropna().unique(), reverse=True)[:10]}")
```

```python
# 3. Transcript 데이터 이상치 확인
print("=== Transcript 이상치 확인 ===")
numeric_cols = ['time']  # Unnamed: 0는 인덱스이므로 제외

for col in numeric_cols:
    if col in transcript.columns:
        print(f"\n[{col}]")
        print(transcript[col].describe())
        print(f"최소값: {transcript[col].min()}, 최대값: {transcript[col].max()}")
        print(f"고유값 개수: {transcript[col].nunique()}")
        
        # time의 경우 음수나 비정상적으로 큰 값 확인
        print(f"\ntime < 0인 행 개수: {(transcript[col] < 0).sum()}")
        
        # 값 분포 확인 (백분위수)
        print(f"\n값 분포:")
        for p in [0, 1, 5, 10, 25, 50, 75, 90, 95, 99, 100]:
            print(f"  {p}%: {transcript[col].quantile(p/100)}")
        
# event 컬럼도 함께 확인 (범주형)
print("\n\n[event - 범주형 변수 확인]")
print(transcript['event'].value_counts())
print(f"\n고유값: {transcript['event'].unique()}")
```

## 결측치 및 이상치 제외 후 데이터 Shape 확인


```python
# 원본 데이터 Shape
print("=== 원본 데이터 Shape ===")
print(f"Portfolio: {portfolio.shape}")
print(f"Profile: {profile.shape}")
print(f"Transcript: {transcript.shape}")

# 결측치 및 이상치 제외 후 데이터 Shape
print("\n=== 결측치 및 이상치 제외 후 데이터 Shape ===")
profile_clean = profile.dropna(subset=['gender', 'income'])
profile_clean = profile_clean[profile_clean['age'] != 118]
transcript_clean = transcript[transcript['person'].isin(profile_clean['id'])]

print(f"Portfolio: {portfolio.shape} (변경 없음)")
print(f"Profile: {profile.shape} → {profile_clean.shape} (유지율: {len(profile_clean)/len(profile)*100:.2f}%)")
print(f"Transcript: {transcript.shape} → {transcript_clean.shape} (유지율: {len(transcript_clean)/len(transcript)*100:.2f}%)")
```
